```python
data.latitude.fillna(data.latitude.mode()[0],inplace=True)
```

```python
data.last_review.fillna(data.last_review.mode()[0],inplace=True)
```

```python
data.reviews_per_month.fillna(data.reviews_per_month.mode()[0],inplace=True)
```

```python
data.calculated_host_listings_count.fillna(data.calculated_host_listings_count.mode()[0],inplace=True)
```

### Detect outliers

```python
import matplotlib.pyplot as plt
```

```python
data["price"] = pd.to_numeric(data["price"], errors="coerce")
data["minimum_nights"] = pd.to_numeric(data["minimum_nights"], errors="coerce")
data["availability_365"] = pd.to_numeric(data["availability_365"], errors="coerce")
```

```python
columns = ["price", "minimum_nights", "availability_365"]

for col in columns:
    plt.figure(figsize=(8, 4))
    plt.boxplot(data[col].dropna())
    plt.title(f"Boxplot of {col}")
    plt.ylabel(col)
    plt.show()
```
