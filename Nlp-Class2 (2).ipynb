# **data PreProcessing**

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
#lower
# Punc
#stopwords
# Extra Spaces
import string
import re # Regural Expression
from nltk.corpus import stopwords

def preprocessor(text):
  text = text.lower()
  text = "".join([a for a in text if a not in string.punctuation])
  text = " ".join([a for a in text.split() if a not in stopwords.words("english")])
  text = re.sub("\s+"," ",text)
  text = re.sub("\d+","",text)
  return text
```

```python
preprocessor("hellooo 6789")
```

```python
data["reviews"] = data["reviews"].apply(preprocessor)
```
